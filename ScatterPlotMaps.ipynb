```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("Book1.csv")

# Scatter Plot
plt.scatter(data["Sales"], data["Quantity"])

# Labels and Title
plt.xlabel("Sales")
plt.ylabel("Quantity")
plt.title("Scatter Plot: Sales vs Quantity")

# Display the plot
plt.show()
```
